```python
PROJECT_ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents)
                    if (p / "notebooks" / "notebook.ipynb").is_file())
ARTIFACTS_PATH = PROJECT_ROOT / "models" / "saved_work"
manifest = json.loads((ARTIFACTS_PATH / "manifest.json").read_text(encoding="utf-8"))
if sklearn.__version__ != manifest["sklearn"]:
    raise RuntimeError(f"Use scikit-learn {manifest['sklearn']} to load these models.")

def load_artifact(name):
    # Load only these locally generated, trusted pickle files.
    with (ARTIFACTS_PATH / name).open("rb") as stream:
        return cloudpickle.load(stream)

dataset = load_artifact("dataset.pkl")
globals().update(dataset)
preprocessing = load_artifact("preprocessing.pkl")
target_scaler = load_artifact("target_scaler.pkl")
models = {name: load_artifact(f"{name}.pkl") for name in
          ("lin_reg", "tree_reg", "forest_reg", "svm_reg", "final_forest_reg")}
globals().update(models)

housing_pre = dataset["housing_pre"]
housing_prepared = dataset["housing_prepared"]
housing_labels = dataset["housing_labels"]
housing_test_prepared = dataset["housing_test_prepared"]
housing_test_labels = dataset["housing_test_labels"]
housing_prepared_tuned = dataset["housing_prepared_tuned"]
housing_test_prepared_tuned = dataset["housing_test_prepared_tuned"]

lin_reg = models["lin_reg"]
tree_reg = models["tree_reg"]
forest_reg = models["forest_reg"]
svm_reg = models["svm_reg"]
final_forest_reg = models["final_forest_reg"]

final_preprocessing = final_forest_reg.named_steps["preprocessing"]
X_train, y_train = housing_prepared, housing_labels
X_test, y_test = housing_test_prepared, housing_test_labels
X_train_tuned, X_test_tuned = housing_prepared_tuned, housing_test_prepared_tuned
print(f"Loaded {X_train.shape[0]} training rows, {X_test.shape[0]} test rows and {len(models)} fitted models.")
```

```python
def predict_dollars(model_name, raw_features):
    predictions = models[model_name].predict(raw_features)
    if manifest["prediction_targets"][model_name] == "scaled":
        predictions = target_scaler.inverse_transform(np.asarray(predictions).reshape(-1, 1)).ravel()
    return predictions

# Saved models are full pipelines: pass raw features, e.g. housing_test.
# For experiments with new estimators, use X_train/y_train (already preprocessed).
predictions = predict_dollars("final_forest_reg", housing_test)
print("Tuned random forest test RMSE ($):", root_mean_squared_error(y_test, predictions))
```
